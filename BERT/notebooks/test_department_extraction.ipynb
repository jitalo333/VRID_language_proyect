```python
import os
import pandas as pd

#load new data
path = "C:/Users/Alonso/Desktop/CDIA_NLP/new_data/"
filepath = os.path.join(path, "data_concat_new.xlsx")

ndf = pd.read_excel(filepath, dtype=str)
ndf = ndf[['Código VRID', 'Tipo Proyecto', 'Interdisciplinario',
    'Transdisciplinario', 'Título', 'Keywords', 'Resumen',
    'Facultad del Proyecto', 'Depto. del Proyecto', 'Año Inicio', 'Depto Persona']]


# AGRUPAR DEPARTAMENTOS POR CÓDIGO VRID
#ndf['Código VRID'] = ndf['Código VRID'].astype(str)
grouped = ndf.groupby('Código VRID')
#Almacenar los depto persona únicos por código VRID
depto_persona = []
for group in grouped:
    unique_deptos = group[1]["Depto Persona"].unique() 
    depto_persona.append(unique_deptos)

#Convertir arrays de lista en strings
depto_persona = [", ".join(deptos) for deptos in depto_persona]
# Quedarse con el primer elemento de cada grupo en el dataframe original
ndf = ndf.drop_duplicates(subset=['Código VRID'], keep='first').copy()
ndf['Depto Persona'] = depto_persona
#Guardar el dataframe modificado
#path = "C:/Users/Alonso/Desktop/CDIA_NLP/new_data/"
#filepath = os.path.join(path, "data_concat_new_final.xlsx")
#ndf.to_excel(filepath, index=False)
```

```python
#load new data
path = "C:/Users/Alonso/Desktop/CDIA_NLP/new_data/"
filepath = os.path.join(path, "data_concat_new.xlsx")

ndf = pd.read_excel(filepath, dtype=str)
ndf_final = (
    ndf.groupby('Código VRID', as_index=False)
       .agg({
           'Tipo Proyecto': 'first',
           'Interdisciplinario': 'first',
           'Transdisciplinario': 'first',
           'Título': 'first',
           'Keywords': 'first',
           'Resumen': 'first',
           'Facultad del Proyecto': 'first',
           'Depto. del Proyecto': 'first',
           'Año Inicio': 'first',
           'Depto Persona': lambda x: ", ".join(sorted(set(x.dropna())))
       })
)

# Guardar el dataframe final
filepath = os.path.join(path, "data_concat_new_final.xlsx")
ndf_final.to_excel(filepath, index=False)
```
