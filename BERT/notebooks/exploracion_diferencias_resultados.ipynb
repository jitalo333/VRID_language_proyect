```python
import json
import pandas as pd
import os

#Ruta de lectura
path = r"C:\Users\Alonso\Desktop\CDIA_NLP\connect_deephub\results_data\data_v9"

#Lectura de data
filepath=os.path.join(path, "data_translated_concat.xlsx")
df_old = pd.read_excel(filepath)

#Ruta de lectura
path = r"C:\Users\Alonso\Desktop\CDIA_NLP\final_proyecto"

#Lectura de data
filepath=os.path.join(path, "data_translated.csv")
df_new = pd.read_csv(filepath)
```

```python
codigos_faltantes = df_old.loc[~df_old["Código VRID"].isin(df_new["Código VRID"])]

print(codigos_faltantes)
```

```python
df_merged = df_old.merge(
    df_new[["Código VRID", "Titulo_trad", "Resumen_trad", "keywords_trad"]],
    on="Código VRID",
    how="left",
    suffixes=("_old", "_new")
)
# Filtrar donde alguna de las columnas no coincide
diferencias = df_merged[
    (df_merged["Titulo_trad_old"] != df_merged["Titulo_trad_new"]) |
    (df_merged["Resumen_trad_old"] != df_merged["Resumen_trad_new"]) |
    (df_merged["keywords_trad_old"] != df_merged["keywords_trad_new"])
]

diferencias.head()
```
