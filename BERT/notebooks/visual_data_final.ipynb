```python
def compare_dataframes(df0, df1):
    # Aseguramos que ambas columnas son string (por si hay mezcla de tipos)
    codes0 = df0["Código VRID"].astype(str).unique()
    codes1 = df1["Código VRID"].astype(str).unique()

    # Convertimos a conjuntos
    set0 = set(codes0)
    set1 = set(codes1)

    # Intersección (elementos que se repiten)
    common_codes = set0.intersection(set1)
    #Codigos que no se repiten
    unique_codes = set0.symmetric_difference(set1)

    print("Cantidad de códigos únicos en df_0:", len(set0))
    print("Cantidad de códigos únicos en df_1:", len(set1))
    print("Cantidad de códigos únicos que se repiten en ambas:", len(common_codes))
    print("Cantidad de códigos que no se repiten:", len(unique_codes))

    # Si además quieres ver cuáles son
    print("Códigos repetidos:", common_codes)

    # Están en df_new0 pero no en df_new1
    only_in_0 = set0 - set1

    # Están en df_new1 pero no en df_new0
    only_in_1 = set1 - set0

    print("Solo en df_new0:", len(only_in_0))
    print("Solo en df_new1:", len(only_in_1))
```

# Evaluar cuantos OCDE y ODS hay en el dataset

```python
path = r"C:\Users\Alonso\Desktop\CDIA_NLP\analysis_last_data"
filename = "new_data.xlsx"

#Pagina 0
df_ocde0 = pd.read_excel(os.path.join(path, filename), sheet_name=2)

#Pagina 1
df_ocde1 =  pd.read_excel(os.path.join(path, filename), sheet_name=3)

df_ocde_new = pd.concat([df_ocde0, df_ocde1])

#Codes unique
codes_unique = df_ocde_new["Código VRID"].astype(str).unique()
print(codes_unique.shape)
```

```python
df_ocde_new["Tipo Área Disciplina"].value_counts()
```

```python
# Ejemplo: df es tu DataFrame
# Contar códigos únicos asociados a ODS
ods_unicos = df_ocde_new.loc[df_ocde_new["Tipo Área Disciplina"] == "ODS", "Código VRID"].nunique()

# Contar códigos únicos asociados a OCDE
ocde_unicos = df_ocde_new.loc[df_ocde_new["Tipo Área Disciplina"] == "OCDE", "Código VRID"].nunique()

print("Cantidad de códigos únicos con ODS:", ods_unicos)
print("Cantidad de códigos únicos con OCDE:", ocde_unicos)
```

```python
duplicados = (
    df_ocde_new.groupby("Código VRID")["Tipo Área Disciplina"]
      .nunique() > 1
)

print("¿Existen códigos con más de un ODS/OCDE?:", duplicados.any())
```

```python
# Agrupamos por Código y Tipo (ODS u OCDE)
conteo = (
    df_ocde_new.groupby(["Código VRID", "Tipo Área Disciplina"])
      .size()
      .reset_index(name="n")
)

# Contamos cuántos ODS distintos tiene cada Código VRID
ods_mult = (
    df_ocde_new[df_ocde_new["Tipo Área Disciplina"] == "ODS"]
    .groupby("Código VRID")["Tipo Área Disciplina"]
    .count()
)
ods_mult = ods_mult[ods_mult > 1]

# Contamos cuántos OCDE distintos tiene cada Código VRID
ocde_mult = (
    df_ocde_new[df_ocde_new["Tipo Área Disciplina"] == "OCDE"]
    .groupby("Código VRID")["Tipo Área Disciplina"]
    .count()
)
ocde_mult = ocde_mult[ocde_mult > 1]

print("Cantidad de códigos con más de un ODS:", len(ods_mult))
print("Cantidad de códigos con más de un OCDE:", len(ocde_mult))
```

```python
df = df_ocde_new.copy()
df = df[df["Categoria Disciplina"] != "SIN AREA"]
# Contar cuántas categorías distintas tiene cada Código VRID
categorias_por_codigo = (
    df.groupby("Código VRID")["Categoria Disciplina"]
      .nunique()
)

# Códigos con más de una categoría distinta asignada
codigos_conflictivos = categorias_por_codigo[categorias_por_codigo > 1]

print("Cantidad de códigos con más de una categoría distinta:",
      len(codigos_conflictivos))

# Opcional: ver cuáles son esos códigos
print(codigos_conflictivos.index.tolist())

print(df["Categoria Disciplina"].value_counts())
```
